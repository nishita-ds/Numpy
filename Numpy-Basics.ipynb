# Stacking and Splitting

```python
# horizontal stacking
a4=np.arange(12).reshape(3,4)
a5=np.arange(12,24).reshape(3,4)
print(a4)
print(a5)
```

```python
# horizontal stacking
b=np.hstack((a4,a5))
```

```python
# vertical stacking
c=np.vstack((a4,a5))
```

```python
# splitting
# horizontal split
np.hsplit(b,2)
```

```python
# vertical split
np.vsplit(c,2)
```
